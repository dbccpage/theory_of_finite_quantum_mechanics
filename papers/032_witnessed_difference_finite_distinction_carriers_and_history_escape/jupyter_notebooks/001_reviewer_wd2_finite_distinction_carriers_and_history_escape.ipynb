## Check the word order by hand

Take states 0, 1, 2, 3 with profiles (0, 0, 0, 1), and updates a = (1, 2, 2, 3), b = (0, 0, 3, 3), listed in state order.

Starting from 0 and 1, applying a then b gives 0 and 3: the profiles differ. Applying b then a gives 1 and 1: the profiles agree. Both single-letter readouts agree. Thus ab is a shortest separating word; reversing it changes the result.
