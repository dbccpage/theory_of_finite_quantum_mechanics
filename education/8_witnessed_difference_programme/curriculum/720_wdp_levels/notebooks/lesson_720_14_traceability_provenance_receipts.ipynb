## 1. Primary Invariants & Discrete Execution: Level 14: Authority is a Separate Layer

```python
# Level 14: Authority Layer & Provenance Receipt
import hashlib


class AuthorityPacket:
    def __init__(self, grantor, scope, key):
        self.grantor = grantor
        self.scope = scope
        self.sig = hashlib.sha256(f"{grantor}:{scope}:{key}".encode()).hexdigest()


auth = AuthorityPacket("Council_L3", "READ_ONLY", "secret_key")
assert len(auth.sig) == 64
print(f"[Level 14 OK] Authority packet signed cleanly: Sig={auth.sig[:12]}...")
```
