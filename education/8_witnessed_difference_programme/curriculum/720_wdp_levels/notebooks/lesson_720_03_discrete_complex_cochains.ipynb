## 1. Primary Invariants & Discrete Coboundary Nilpotency

Verify that d1(d0 f) = 0 for an arbitrary rational vertex cochain f.


```python
from fractions import Fraction

# Arbitrary rational vertex cochain on triangle (0, 1, 2)
f = {0: Fraction(13, 7), 1: Fraction(-5, 3), 2: Fraction(42, 11)}

# Compute d0 f on oriented edges (0,1), (1,2), (0,2)
d0_f = {
    (0, 1): f[1] - f[0],
    (1, 2): f[2] - f[1],
    (0, 2): f[2] - f[0],
}

# Compute d1(d0 f) on face (0, 1, 2)
curl = d0_f[(0, 1)] + d0_f[(1, 2)] - d0_f[(0, 2)]
assert curl == Fraction(0), f"Nilpotency violation: curl={curl}"
print("[Level 03 OK] d1(d0 f) = 0 exactly verified on rational vertex inputs.")
```

## 3. Diagnostic Probes & Adversarial Falsifiers

Verify that a non-gradient cochain produces a non-zero curl on a declared face.


```python
# Falsifier test: Non-gradient cochain with non-zero curl
h_curl = {
    (0, 1): Fraction(1),
    (1, 2): Fraction(1),
    (0, 2): Fraction(0),
}
curl_nonzero = h_curl[(0, 1)] + h_curl[(1, 2)] - h_curl[(0, 2)]
assert curl_nonzero == Fraction(2)
print(f"[Falsifier OK] Non-gradient cochain correctly yields non-zero curl: {curl_nonzero}")
```
