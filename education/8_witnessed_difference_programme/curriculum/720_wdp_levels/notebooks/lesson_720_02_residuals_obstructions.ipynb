## 1. Primary Invariants & Residual Computation

Verify the calculation of residual vector r = delta - d0 f on a directed graph.


```python
from fractions import Fraction

# Triangle with edge cochain delta and vertex potential f
edges = ((0, 1), (1, 2), (0, 2))
delta = (Fraction(1), Fraction(1), Fraction(0))
potential = {0: Fraction(0), 1: Fraction(1), 2: Fraction(2)}

# Compute gradient (d0 f)(u, v) = f(v) - f(u)
grad = tuple(potential[head] - potential[tail] for tail, head in edges)
assert grad == (Fraction(1), Fraction(1), Fraction(2))

# Compute residual r = delta - grad
residual = tuple(d - g for d, g in zip(delta, grad))
assert residual == (Fraction(0), Fraction(0), Fraction(-2))

# L1 norm of residual
residual_norm = sum(abs(r) for r in residual)
assert residual_norm == Fraction(2)
print(f"[Level 02 OK] Residual vector computed: {residual}, norm={residual_norm}")
```

## 3. Diagnostic Probes & Adversarial Falsifiers

Verify that an exact potential eliminates the residual norm to zero.


```python
# Falsifier test: Exact cochain matching gradient achieves zero residual
exact_delta = (Fraction(1), Fraction(1), Fraction(2))
exact_potential = {0: Fraction(0), 1: Fraction(1), 2: Fraction(2)}
exact_grad = tuple(exact_potential[head] - exact_potential[tail] for tail, head in edges)
exact_residual = tuple(d - g for d, g in zip(exact_delta, exact_grad))
assert sum(abs(r) for r in exact_residual) == Fraction(0)
print("[Falsifier OK] Exact gradient achieves zero residual norm.")
```
