## 1. Primary Invariants & Discrete Execution: Level 08: Probability Scope Fences

Execute verified discrete calculations and assertions for Level specifications.


```python
# Level 08: Probability Interpretation Scope Fences
from enum import Enum, auto


class ProbType(Enum):
    COUNTING = auto()
    FREQUENCY = auto()
    PROPENSITY = auto()
    BAYESIAN = auto()
    BORN_WEIGHT = auto()


class TypedProbability:
    def __init__(self, value, prob_type: ProbType):
        self.value = value
        self.prob_type = prob_type


p_count = TypedProbability(0.5, ProbType.COUNTING)
p_born = TypedProbability(0.5, ProbType.BORN_WEIGHT)

assert p_count.prob_type != p_born.prob_type, "Scope fence failed"
print("[Level 08 OK] Probability scope fence strictly enforced between Counting and Born weights")
```

## 2. Diagnostic Probes & Adversarial Falsifiers

Verify that un-admitted, unauthorized, or invalid operations are intercepted and rejected.


```python
# Falsifier test: Attempting implicit type coercion between probability types must fail
def coerce_to_born(typed_p: TypedProbability):
    if typed_p.prob_type != ProbType.BORN_WEIGHT:
        raise TypeError(f"[Falsifier OK] Cannot coerce {typed_p.prob_type.name} to BORN_WEIGHT")


try:
    coerce_to_born(p_count)
except TypeError as e:
    print(e)
```
