## 1. Primary Invariants & The Diagnostic Trichotomy

Verify the three mutually exclusive diagnostic states on discrete complexes.


```python
from enum import Enum
from fractions import Fraction


class DiagnosticClass(str, Enum):
    EXACT_CLASS = "exact_class"
    CLOSED_RESIDUAL = "closed_residual"
    ACTIVE_CLOSURE_FAILURE = "active_closure_failure"


def classify_signature(phi: Fraction, gamma: Fraction) -> DiagnosticClass:
    if gamma > 0:
        return DiagnosticClass.ACTIVE_CLOSURE_FAILURE
    if phi > 0:
        return DiagnosticClass.CLOSED_RESIDUAL
    return DiagnosticClass.EXACT_CLASS


# 1. Frustrated closed loop without faces: phi > 0, gamma == 0
assert classify_signature(Fraction(2), Fraction(0)) is DiagnosticClass.CLOSED_RESIDUAL

# 2. Active boundary failure on declared face: gamma > 0
assert classify_signature(Fraction(2), Fraction(1)) is DiagnosticClass.ACTIVE_CLOSURE_FAILURE

# 3. Exact potential gradient: phi == 0, gamma == 0
assert classify_signature(Fraction(0), Fraction(0)) is DiagnosticClass.EXACT_CLASS

print("[Trichotomy OK] All three mutually exclusive diagnostic states verified.")
```

## 3. Diagnostic Probes & Adversarial Falsifiers

Verify that a signature cannot simultaneously belong to multiple classes.


```python
# Falsifier test: Exhaustive classification coverage across non-negative rational pairs
test_pairs = [
    (Fraction(0), Fraction(0)),
    (Fraction(1), Fraction(0)),
    (Fraction(0), Fraction(1)),
    (Fraction(3, 2), Fraction(5, 7)),
]
for phi, gamma in test_pairs:
    cls = classify_signature(phi, gamma)
    # Exactly one condition holds
    is_active = gamma > 0
    is_residual = gamma == 0 and phi > 0
    is_exact = gamma == 0 and phi == 0
    assert sum([is_active, is_residual, is_exact]) == 1

print("[Falsifier OK] Zero diagnostic class collision or ambiguity.")
```
