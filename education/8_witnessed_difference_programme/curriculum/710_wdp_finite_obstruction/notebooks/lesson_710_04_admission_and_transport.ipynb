```python
import sys
from fractions import Fraction as Q

assert sys.version_info >= (3, 10)
print("Exact finite examples; AuthorityEffect: None")
```

```python
# Commit predictions before running the checks. Change these to explore.
prediction = {
    "law_without_exhibit": False,
    "reverse_cut": True,
    "homotopy_all_raw_classes": False,
    "up_down_identity": False,
}
print(prediction)
```

```python
# A finite table defines R exactly on its declared carrier.
def checked_cut(carrier, relation, exhibit):
    if len(set(carrier)) != len(carrier):
        raise ValueError("duplicate carrier")
    if any(a not in carrier or b not in carrier for a, b in relation):
        raise ValueError("endpoint outside carrier")
    if any(a == b for a, b in relation):
        raise ValueError("irreflexivity fails")
    if exhibit not in relation:
        raise ValueError("missing exhibit")
    a, b = exhibit
    assert a != b
    return True


assert checked_cut(("a", "b"), {("a", "b")}, ("a", "b"))
reverse = checked_cut(("a", "b"), {("b", "a")}, ("b", "a"))
assert reverse == prediction["reverse_cut"]
for carrier, relation, exhibit in [
    (("a",), {("a", "a")}, ("a", "a")),
    (("a",), set(), ("a", "a")),
    (("a",), {("a", "outside")}, ("a", "outside")),
]:
    rejected = False
    try:
        checked_cut(carrier, relation, exhibit)
    except ValueError:
        rejected = True
    assert rejected
assert prediction["law_without_exhibit"] is False
```

## Keep the quotient question fixed
The next functions illustrate the native test complex `(1,2,1)`.
The repair class is the second coordinate; closed elements have that coordinate zero.
This lab contains no elimination algorithm or production solver.


```python
def d0(t):
    return (t, Q(0))


def d1(x):
    return x[1]


def add(x, y):
    return tuple(a + b for a, b in zip(x, y))


def raw_class(x):
    return x[1]


assert all(d1(d0(t)) == 0 for t in (Q(-2), Q(0), Q(3, 2)))
x = (Q(3), Q(1))
assert raw_class(add(x, d0(Q(7)))) == raw_class(x)
assert d1((Q(3), Q(0))) == 0
assert d1((Q(0), Q(1))) != 0
```

```python
def h1(x):
    return x[0]


def h2(t):
    return (Q(0), t)


def filler(x):
    return add(d0(h1(x)), h2(d1(x)))


samples = [(Q(3), Q(0)), (Q(0), Q(1)), (Q(2, 3), Q(-4, 5))]
assert all(filler(x) == x for x in samples)
closed = (Q(3), Q(0))
nonclosed = (Q(0), Q(1))
zero = (Q(0), Q(0))
assert raw_class(closed) == raw_class(zero)
assert raw_class(nonclosed) != raw_class(zero)
assert prediction["homotopy_all_raw_classes"] is False
print("Homotopy agrees on closed classes; a nonclosed raw class survives.")
```

```python
# Malformed filler: H2(t)=(0,2t) breaks the middle equation.
def false_filler(x):
    return add(d0(h1(x)), (Q(0), 2 * d1(x)))


assert false_filler(nonclosed) != nonclosed


# A wrong differential similarly invalidates the complex itself.
def false_d1(x):
    return x[0] + x[1]


assert false_d1(d0(Q(1))) != 0
```

```python
def up(t):
    return (t, Q(0))


def down(x):
    return x[0]


assert all(down(up(t)) == t for t in (Q(-1), Q(0), Q(3, 4)))
assert up(down(nonclosed)) != nonclosed
assert prediction["up_down_identity"] is False
print("Retraction preserves the old image; the new direction remains distinct.")
```
