## 1. The Six Anti-Cheat Halting Laws

Under `omega_wdp::prime_algos::licensed_lift_gate`, the following six laws govern every ascent move:
1. **Law 1 (No Unclosed Lift)**: Cannot lift when curl $\Gamma_2 > 0$.
2. **Law 2 (Zero Defect Halt)**: If $\Phi_1 = 0$, state is already exact; lift is forbidden.
3. **Law 3 (Strict Lexicographic Dominance)**: Ascent step must decrease defect or record certified refusal.
4. **Law 4 (Cycle Bound)**: Repeated visit of an identical state hash triggers immediate halt.
5. **Law 5 (Licensed Context Exclusivity)**: Carrier extension requires explicit non-null context contract.
6. **Law 6 (Authority Prohibition)**: Ascent receipts carry `AuthorityEffect: None`.


```python
halting_laws = [
    "Law 1: No Unclosed Lift (Gamma2 == 0 required)",
    "Law 2: Zero Defect Halt (Phi1 > 0 required)",
    "Law 3: Strict Lexicographic Dominance (defect reduction)",
    "Law 4: Finite Memory / Cycle Bound (no repeated state hash)",
    "Law 5: Licensed Context Exclusivity (authenticated context contract)",
    "Law 6: Non-Authority Invariant (AuthorityEffect: None)",
]
assert len(halting_laws) == prediction["halting_laws_count"]
print(f"[Laws OK] All {len(halting_laws)} halting laws registered.")
```

## 2. Fail-Closed Halting Law Evaluator

Implement the gate evaluating candidate ascent steps against the halting laws:


```python
def evaluate_ascent_move(phi1, gamma2, context_license, state_history, current_state_hash):
    # Law 1
    if gamma2 > 0:
        return False, "REFUSED_UNCLOSED"
    # Law 2
    if phi1 == 0:
        return False, "REFUSED_EXACT"
    # Law 4
    if current_state_hash in state_history:
        return False, "REFUSED_CYCLE_DETECTED"
    # Law 5
    if not context_license or not context_license.get("valid", False):
        return False, "REFUSED_NO_LICENSE"
    # Passed
    return True, "ASCENT_PERMITTED"

# Test Law 1
ok1, reason1 = evaluate_ascent_move(Fraction(1, 1), Fraction(1, 1), {"valid": True}, set(), "h1")
assert not ok1 and reason1 == prediction["unclosed_lift_refusal"], f"Law 1 failed: {reason1}"

# Test Law 2
ok2, reason2 = evaluate_ascent_move(Fraction(0, 1), Fraction(0, 1), {"valid": True}, set(), "h2")
assert not ok2 and reason2 == prediction["zero_defect_lift_refusal"], f"Law 2 failed: {reason2}"

# Test Law 4
history = {"h_visited"}
ok4, reason4 = evaluate_ascent_move(Fraction(1, 1), Fraction(0, 1), {"valid": True}, history, "h_visited")
assert not ok4 and reason4 == prediction["cycle_detection_refusal"], f"Law 4 failed: {reason4}"

# Test Valid Step
ok_valid, reason_valid = evaluate_ascent_move(Fraction(1, 1), Fraction(0, 1), {"valid": True}, history, "h_new")
assert ok_valid and reason_valid == "ASCENT_PERMITTED"

print("[Evaluator OK] Halting laws 1, 2, 4, 5 verified fail-closed.")
```

## 3. The Nine Up-Lift Gate Requirements

Verify that the 9 prerequisite conditions for carrier expansion are strictly checked:


```python
class UpLiftGate:
    def __init__(self):
        self.checks = [
            "closed_cocycle",
            "non_exact_residual",
            "admitted_license",
            "loss_ledger_reconciled",
            "bounded_dimension_limit",
            "acyclic_trajectory",
            "zero_curl",
            "capacity_available",
            "authority_none",
        ]

    def verify_prerequisites(self, state_dict):
        for check in self.checks:
            if not state_dict.get(check, False):
                return False, f"Gate failed on check: {check}"
        return True, "ALL_9_PREREQUISITES_PASSED"

gate = UpLiftGate()
assert len(gate.checks) == 9

valid_state = {c: True for c in gate.checks}
passed, msg = gate.verify_prerequisites(valid_state)
assert passed and msg == "ALL_9_PREREQUISITES_PASSED"

# Perturb one check
invalid_state = dict(valid_state)
invalid_state["loss_ledger_reconciled"] = False
failed, fail_msg = gate.verify_prerequisites(invalid_state)
assert not failed and "loss_ledger_reconciled" in fail_msg
print("[UpLiftGate OK] Nine Up-Lift Gate verified.")
```
