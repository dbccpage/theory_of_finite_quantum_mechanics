## 1. Primary Invariants & The Outcome Monad

Verify that the outcome monad distinguishes Accepted from Refused with typed codes.


```python
from dataclasses import dataclass
from enum import Enum


class AuthorityEffect(str, Enum):
    NONE = "None"


class RefusalCode(str, Enum):
    INVALID_INPUT = "invalid_input"
    SCOPE_VIOLATION = "scope_violation"


@dataclass(frozen=True)
class Accepted:
    value: object
    authority_effect: AuthorityEffect = AuthorityEffect.NONE


@dataclass(frozen=True)
class Refused:
    code: RefusalCode
    detail: str
    authority_effect: AuthorityEffect = AuthorityEffect.NONE


acc = Accepted(42)
assert acc.value == 42
assert acc.authority_effect is AuthorityEffect.NONE
print("[Monad OK] Accepted outcome verified with AuthorityEffect.NONE")
```

## 3. Diagnostic Probes & Adversarial Falsifiers

Verify that invalid inputs generate a typed Refused outcome rather than unhandled exceptions.


```python
def safe_divide(numerator: int, denominator: int):
    if denominator == 0:
        return Refused(RefusalCode.INVALID_INPUT, "Zero division not permitted")
    return Accepted(numerator // denominator)


res_ok = safe_divide(10, 2)
assert isinstance(res_ok, Accepted) and res_ok.value == 5

res_err = safe_divide(10, 0)
assert isinstance(res_err, Refused)
assert res_err.code is RefusalCode.INVALID_INPUT
print("[Falsifier OK] Zero division safely contained in Refused monad.")
```
