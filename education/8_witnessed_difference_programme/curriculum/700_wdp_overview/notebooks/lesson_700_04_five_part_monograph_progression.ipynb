## 1. Primary Invariants & Five-Part Architecture

Verify the chapter boundaries across Parts I through V.


```python
parts = {
    "Part I": tuple(range(1, 6)),  # Ch 1-5: Difference & Obstruction
    "Part II": tuple(range(6, 10)),  # Ch 6-9: Order & Probability
    "Part III": tuple(range(10, 16)),  # Ch 10-15: Observers & Measurement
    "Part IV": tuple(range(16, 20)),  # Ch 16-19: Histories & Quantum Boundary
    "Part V": tuple(range(20, 23)),  # Ch 20-22: Scope & Open Horizon
}

all_chapters = [ch for part in parts.values() for ch in part]
assert len(all_chapters) == 22
assert all_chapters == list(range(1, 23))
print("[Architecture OK] 22 chapters partitioned across 5 parts without gaps.")
```

## 3. Diagnostic Probes & Adversarial Falsifiers

Verify that chapters cannot overlap or leak between parts.


```python
# Falsifier test: No chapter is duplicated across parts
seen = set()
for part_name, ch_list in parts.items():
    for ch in ch_list:
        assert ch not in seen, f"Chapter {ch} leaked across parts!"
        seen.add(ch)
print("[Falsifier OK] Zero cross-part chapter duplication.")
```
