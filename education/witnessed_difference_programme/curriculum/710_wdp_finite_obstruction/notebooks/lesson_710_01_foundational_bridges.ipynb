## 1. Primary Invariants & Coboundary Nilpotency (Bridge 2 & 9)

Verify that the composite operator satisfies $d_1 \circ d_0 = 0$ exactly over rational coefficients.


```python
# Construct boundary matrices for triangle K3: vertices (0, 1, 2), edges ((0,1), (1,2), (0,2)), face (0,1,2)
# d0: edges x vertices
d0 = [
    [-1, 1, 0],  # (0, 1): v1 - v0
    [0, -1, 1],  # (1, 2): v2 - v1
    [-1, 0, 1],  # (0, 2): v2 - v0
]
# d1: faces x edges -> (0,1) + (1,2) - (0,2)
d1 = [
    [1, 1, -1],
]

# Compute composite d1 o d0
composite = [[sum(d1[f][e] * d0[e][v] for e in range(3)) for v in range(3)] for f in range(1)]

assert composite == prediction["triangle_composite"], f"Nilpotency failure: {composite}"
print("[Bridge 2 & 9 OK] Coboundary nilpotency d1 o d0 = 0 verified.")
```

## 3. Diagnostic Probes & Adversarial Falsifiers (Bridge 10)

Verify that frustrated all-edge anticorrelation fails globally on an odd cycle.


```python
# Test all 2^3 = 8 binary assignments s in {0, 1}^3 on 3-cycle
all_anticorrelated = []
for s0 in (0, 1):
    for s1 in (0, 1):
        for s2 in (0, 1):
            # Condition: s0 ^ s1 == 1 and s1 ^ s2 == 1 and s2 ^ s0 == 1
            if (s0 ^ s1 == 1) and (s1 ^ s2 == 1) and (s2 ^ s0 == 1):
                all_anticorrelated.append((s0, s1, s2))

# Falsifier test: No global binary assignment can satisfy odd-cycle anticorrelation!
assert len(all_anticorrelated) == prediction["anticorrelated_assignments"], (
    "Frustration failure: global anticorrelation admitted"
)
print("[Bridge 10 OK] Odd cycle frustration verified: 0 valid global anticorrelations.")
```
