## 2. Prime 000: 12-Step Obstruction Fill Simulation

Execute a 12-step obstruction fill cycle over a discrete carrier state:


```python
class CarrierState:
    def __init__(self, coordinates):
        self.coordinates = dict(coordinates)
        self.defect_history = []

    def total_energy(self):
        return sum(abs(v) for v in self.coordinates.values())

carrier = CarrierState({f"coord_{i}": Fraction(12 - i, 1) for i in range(12)})
initial_energy = carrier.total_energy()

# Execute 12-step descent fill
for step in range(12):
    key = f"coord_{step}"
    old_val = carrier.coordinates[key]
    carrier.coordinates[key] = Fraction(0, 1)
    carrier.defect_history.append((step, old_val, Fraction(0, 1)))

assert len(carrier.defect_history) == prediction["fill_steps"], "Fill steps mismatch"
final_energy = carrier.total_energy()
assert final_energy == Fraction(0, 1), f"Defect not cleared: {final_energy}"
print(f"[Prime 000 OK] 12-step obstruction fill reduced energy from {initial_energy} to {final_energy}.")
```

## 3. Prime 001: Scale Descent and Loss Ledger Conservation

Coarse-graining a carrier by projection drops high-frequency coordinates. The dropped coordinates must be recorded in the loss ledger such that:
$$\text{dim}(C_{\mathrm{in}}) = \text{dim}(C_{\mathrm{coarse}}) + |\mathcal{L}_{\mathrm{loss}}|$$


```python
class LossLedger:
    def __init__(self):
        self.entries = []

    def record_drop(self, coordinate_id, value):
        self.entries.append({"id": coordinate_id, "value": value})

    def count(self):
        return len(self.entries)

def scale_descent_project(carrier_coords, target_dim):
    coarse = {}
    ledger = LossLedger()
    items = sorted(carrier_coords.items())
    for idx, (coord_id, val) in enumerate(items):
        if idx < target_dim:
            coarse[coord_id] = val
        else:
            ledger.record_drop(coord_id, val)
    return coarse, ledger

source_carrier = {f"basis_{i}": Fraction(i + 1, 1) for i in range(8)}
target_dim = 3
coarse_state, loss_ledger = scale_descent_project(source_carrier, target_dim)

conservation_holds = (len(source_carrier) == len(coarse_state) + loss_ledger.count())
assert conservation_holds == prediction["loss_ledger_conservation"]
print(f"[Prime 001 OK] Scale descent conserved coordinates: {len(source_carrier)} = {len(coarse_state)} + {loss_ledger.count()}")
```

## 4. Prime 002: Context License Gate Verification

A structural lift $\Lambda$ cannot be invoked arbitrarily. It requires an authenticated context contract.


```python
def attempt_structural_lift(carrier_state, context_license):
    if not context_license or context_license.get("status") != "AUTHENTICATED":
        return None, "REFUSED: Missing or invalid context license"
    # Admitted lift
    lifted = dict(carrier_state)
    lifted["extended_basis_0"] = Fraction(0, 1)
    return lifted, "LIFT_GRANTED"

unlicensed_result, reason = attempt_structural_lift(coarse_state, None)
assert unlicensed_result is None
assert "REFUSED" in reason

licensed_result, reason_ok = attempt_structural_lift(coarse_state, {"status": "AUTHENTICATED", "boundary_id": "CTX-750"})
assert licensed_result is not None
assert reason_ok == "LIFT_GRANTED"
print("[Prime 002 OK] Context license firewall successfully gatekeeps structural lift.")
```
