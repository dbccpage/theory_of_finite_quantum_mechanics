## 2. Cochain Complex: Delta Nilpotency and Discrete Operators

Construct a finite cochain complex $C^0 \xrightarrow{d_0} C^1 \xrightarrow{d_1} C^2$ on a simplicial 3-cycle $K_3$ with 3 vertices, 3 edges, and 1 face:
- $C^0 = \mathbb{Q}^3$ (vertex potentials $t = (t_0, t_1, t_2)$)
- $C^1 = \mathbb{Q}^3$ (edge cochains $x = (x_{01}, x_{12}, x_{02})$)
- $C^2 = \mathbb{Q}^1$ (face curl)


```python
# Boundary matrices
# d0: C0 -> C1 (edges x vertices)
d0 = [
    [-1, 1, 0],   # e01: v1 - v0
    [0, -1, 1],   # e12: v2 - v1
    [-1, 0, 1],   # e02: v2 - v0
]

# d1: C1 -> C2 (faces x edges)
d1 = [
    [1, 1, -1],   # f012: e01 + e12 - e02
]

# Check nilpotency d1 o d0 = 0
d1_o_d0 = [[sum(d1[f][e] * d0[e][v] for e in range(3)) for v in range(3)] for f in range(1)]
assert d1_o_d0 == [[0, 0, 0]], f"Nilpotency failure: {d1_o_d0}"
print("[Nilpotency OK] d1 o d0 = 0 confirmed.")
```

## 3. Cochain Diagnostics: $\Phi_1([x])$ Primal Quotient and $\Gamma_2(x)$ Curl Invariant

We compute the exact rational $L_1$ quotient $\Phi_1([x])$ and curl $\Gamma_2(x)$:
$$\Phi_1([x]) = \min_{t \in C^0} \|x - d_0 t\|_1$$
$$\Gamma_2(x) = \|d_1 x\|_1$$


```python
def compute_coboundary(t):
    return [
        -t[0] + t[1],
        -t[1] + t[2],
        -t[0] + t[2],
    ]

def compute_curl(x):
    return abs(x[0] + x[1] - x[2])

def phi1_quotient_bruteforce(x, candidate_grid):
    best_dist = None
    best_t = None
    for t in candidate_grid:
        d0_t = compute_coboundary(t)
        dist = sum(abs(x[i] - d0_t[i]) for i in range(3))
        if best_dist is None or dist < best_dist:
            best_dist = dist
            best_t = t
    return best_dist, best_t

# Grid over rational potential candidates
grid = [
    (Fraction(v0, 1), Fraction(v1, 1), Fraction(v2, 1))
    for v0 in range(-2, 3)
    for v1 in range(-2, 3)
    for v2 in range(-2, 3)
]

# Case 1: Pure Coboundary
t_exact = (Fraction(1, 1), Fraction(2, 1), Fraction(4, 1))
x_coboundary = compute_coboundary(t_exact)
phi1_val, t_found = phi1_quotient_bruteforce(x_coboundary, grid)
gamma2_val = compute_curl(x_coboundary)

assert phi1_val == prediction["phi1_coboundary"], f"Expected 0, got {phi1_val}"
assert gamma2_val == Fraction(0, 1), f"Expected 0 curl, got {gamma2_val}"
print(f"[Case 1 OK] Pure coboundary: Phi_1 = {phi1_val}, Gamma_2 = {gamma2_val}")

# Case 2: Open cochain (curl defect)
x_open = [Fraction(1, 1), Fraction(0, 1), Fraction(0, 1)]
gamma2_open = compute_curl(x_open)
assert gamma2_open == prediction["gamma2_curl_defect"], f"Expected 1, got {gamma2_open}"
print(f"[Case 2 OK] Open cochain: Gamma_2 = {gamma2_open}")
```
