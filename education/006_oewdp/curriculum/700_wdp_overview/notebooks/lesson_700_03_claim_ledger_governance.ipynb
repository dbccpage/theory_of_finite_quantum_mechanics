## 1. Primary Invariants & Claim Hygiene Taxonomy

Verify the presence of forbidden words and enforce replacement posture.


```python
forbidden_words = (
    "proves",
    "guarantees",
    "mathematically verifies",
    "fundamentally",
    "true intelligence",
    "impossible",
    "universal",
    "flawless",
    "obviously",
    "simply",
)

sample_claim = "The finite test supports correctness under the declared assumptions."
assert not any(word in sample_claim.lower().split() for word in forbidden_words)
print("[Hygiene OK] Claim conforms to strict claim hygiene rules.")
```

## 2. Prediction Gate

Commit state expectation before executing verification probes.


```python
# Prediction Gate: Commit state expectation
prediction = "HYGIENE_CLEAN"
assert prediction == "HYGIENE_CLEAN", "Prediction mismatch"
print("[Prediction Confirmed] Diagnostic gate: HYGIENE_CLEAN")
```

## 3. Diagnostic Probes & Adversarial Falsifiers

Verify that hype prose is detected and flagged by the linter guard.


```python
hype_text = "This algorithm fundamentally proves universal intelligence."
detected = [w for w in forbidden_words if w in hype_text.lower()]
assert len(detected) >= 2, "Failed to detect forbidden words"
print(f"[Falsifier OK] Intercepted forbidden words: {detected}")
```
