## 1. Disjoint Terminal Factorization Variants

Under `omega_wdp::coda::coda_gtla`, any resolution process terminates in exactly one of three disjoint variants:
- **$T_{\mathrm{exact}}$**: Exact coboundary reduction achieved with dual certificate of zero residual.
- **$T_{\mathrm{survivor}}$**: Closed non-exact harmonic residual admitted to persistent register.
- **$T_{\mathrm{refusal}}$**: Curvature defect, unclosed boundary, or unauthorized extension halted fail-closed.


```python
terminal_variants = ["T_exact", "T_survivor", "T_refusal"]
assert len(terminal_variants) == prediction["terminal_variants"], "Variant count mismatch"

def route_terminal(status_code, phi1, gamma2, authorized_context):
    if gamma2 > 0 or not authorized_context:
        return "T_refusal", "Refusal: non-zero curl or missing context license"
    elif phi1 == 0 and gamma2 == 0:
        return "T_exact", "Exact coboundary: residual vanishes"
    elif phi1 > 0 and gamma2 == 0:
        return "T_survivor", "Survivor: persistent harmonic residual"
    else:
        return "T_refusal", "Fail-closed refusal"

# Test partition exclusivity
outcomes = [
    route_terminal("OK", Fraction(0, 1), Fraction(0, 1), True)[0],
    route_terminal("OK", Fraction(1, 2), Fraction(0, 1), True)[0],
    route_terminal("FAIL", Fraction(1, 2), Fraction(1, 1), True)[0],
    route_terminal("OK", Fraction(0, 1), Fraction(0, 1), False)[0],
]

assert set(outcomes) == set(terminal_variants), "Not all variants realized"
assert len(outcomes) == 4
print("[Partition OK] All 3 disjoint terminal variants exercised.")
```

## 2. Primal-Dual Witness Certification over $\mathbb{Q}$

Consider the $L_1$ primal quotient problem on finite cochains:
$$\Phi_1([x]) = \min_{t \in C^0} \|x - d_0 t\|_1$$
By linear programming duality, the dual problem is:
$$\Phi_1^*([x]) = \max \left\{ \langle y, x \rangle : y \in C^1, \|y\|_\infty \le 1, d_0^T y = 0 \right\}$$
Strong duality guarantees that on finite rational complexes, the duality gap $\Delta = \Phi_1([x]) - \Phi_1^*([x]) = 0$.


```python
# Given cochain x on a 2-edge line graph: v0 -> v1 -> v2
# d0: e01 = v1 - v0, e12 = v2 - v1
# d0^T = [[-1, 0], [1, -1], [0, 1]]
# Condition d0^T y = 0 => -y0 = 0, y0 - y1 = 0, y1 = 0 => y = (0, 0)
# Here im(d0) is full dimension on edges! Every x is coboundary!

x = [Fraction(3, 1), Fraction(5, 1)]
# Solve t: t1 - t0 = 3, t2 - t1 = 5. Set t0 = 0 => t1 = 3, t2 = 8.
t = [Fraction(0, 1), Fraction(3, 1), Fraction(8, 1)]
d0_t = [t[1] - t[0], t[2] - t[1]]
primal_obj = sum(abs(x[i] - d0_t[i]) for i in range(2))

# Dual witness y = (0, 0) since ker(d0^T) = {0}
y = [Fraction(0, 1), Fraction(0, 1)]
dual_obj = sum(y[i] * x[i] for i in range(2))

duality_gap = primal_obj - dual_obj
assert duality_gap == prediction["duality_gap"], f"Duality gap non-zero: {duality_gap}"
print(f"[Duality OK] Primal = {primal_obj}, Dual = {dual_obj}, Duality Gap = {duality_gap}")
```
