## 1. Primary Invariants & Discrete Execution: Level 07: Multiplicity, Additive Weight, and Finite Probability

Execute verified discrete calculations and assertions for Level specifications.


```python
# Level 07: Multiplicity & Exact Rational Probability
from fractions import Fraction

outcomes = {"a": 2, "b": 3, "c": 5}
total_weight = sum(outcomes.values())
probabilities = {k: Fraction(v, total_weight) for k, v in outcomes.items()}

assert sum(probabilities.values()) == Fraction(1, 1), "Probability normalization error"
assert probabilities["c"] == Fraction(1, 2), "Ratio calculation error"
print(f"[Level 07 OK] Exact rational probabilities verified: {probabilities}")
```
