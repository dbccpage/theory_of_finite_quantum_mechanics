## 1. Primary Invariants & Discrete Execution: Level 06: Duration, Exactness, Periods, and Clocks

Execute verified discrete calculations and assertions for Level specifications.


```python
# Level 06: Additive Event Valuation & Clock Periods
from fractions import Fraction

edge_weights = {(0, 1): Fraction(1, 2), (1, 2): Fraction(1, 3), (0, 2): Fraction(5, 6)}
path1_dur = edge_weights[(0, 1)] + edge_weights[(1, 2)]
path2_dur = edge_weights[(0, 2)]

assert path1_dur == path2_dur == Fraction(5, 6), "Path duration mismatch"
print(f"[Level 06 OK] Exact rational path duration verified: {path1_dur}")
```

```python
# Falsifier test: Non-zero cycle period breaks path independence
cycle_weights = {(0, 1): Fraction(1, 2), (1, 2): Fraction(1, 2), (2, 0): Fraction(1, 3)}
cycle_period = sum(cycle_weights.values())
assert cycle_period != Fraction(0, 1), "Cycle valuation should be non-zero"
print(f"[Falsifier OK] Cyclic clock period obstruction detected: {cycle_period}")
```
