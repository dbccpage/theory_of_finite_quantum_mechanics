## 1. Primary Invariants & Witness Receipts

Verify that a witness receipt carries immutable data and zero authority effect.


```python
from dataclasses import dataclass
from enum import Enum


class AuthorityEffect(str, Enum):
    NONE = "None"


@dataclass(frozen=True)
class WitnessReceipt:
    subject: str
    evidence: str
    authority_effect: AuthorityEffect = AuthorityEffect.NONE


receipt = WitnessReceipt("CARRIER-01", "EXPLICIT_EVIDENCE")
assert receipt.authority_effect is AuthorityEffect.NONE
print("[Level 01 OK] Witness receipt instantiated with AuthorityEffect.NONE")
```

Commit state expectation before executing verification probes.

## 3. Diagnostic Probes & Adversarial Falsifiers

Verify that an attempt to mutate a frozen receipt fails closed.


```python
# Falsifier test: Frozen dataclass must reject attribute assignment
mutation_rejected = False
try:
    receipt.authority_effect = "AUTHORITATIVE"  # type: ignore
except Exception:
    mutation_rejected = True

assert mutation_rejected, "Frozen witness receipt permitted mutation!"
print("[Falsifier OK] Illegal receipt mutation rejected.")
```
