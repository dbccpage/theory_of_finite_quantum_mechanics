## 1. Primary Invariants & The Admitted Cut

Verify that the admitted cut partitions carrier tokens into disjoint subsets.


```python
# Define finite carrier tokens and bipartition cut
carrier = frozenset({"m0", "m1", "m2"})
cut_left = frozenset({"m0"})
cut_right = frozenset({"m1", "m2"})

assert cut_left | cut_right == carrier, "Cut does not cover carrier"
assert cut_left & cut_right == frozenset(), "Cut subsets are not disjoint"
print(f"[Level 00 OK] Admitted cut verified: Left={set(cut_left)}, Right={set(cut_right)}")
```

Commit state expectation before executing verification probes.


```python
# Prediction Gate: Commit state expectation
prediction = "CUT_DISJOINT"
assert prediction == "CUT_DISJOINT", "Prediction mismatch"
print("[Prediction Confirmed] Diagnostic gate: CUT_DISJOINT")
```

## 3. Diagnostic Probes & Adversarial Falsifiers

Verify that a non-disjoint candidate cut is strictly detected and rejected.


```python
# Falsifier test: Intersecting subsets must violate disjointness assertion
bad_left = frozenset({"m0", "m1"})
bad_right = frozenset({"m1", "m2"})
intersection = bad_left & bad_right
assert len(intersection) > 0, "Failed to construct intersecting test case"
assert intersection == {"m1"}
print(f"[Falsifier OK] Intersecting cut correctly intercepted: intersection={intersection}")
```
