## 1. Primary Invariants & Discrete Execution: Level 12: Measurement as a Typed Extension of Observation

Execute verified discrete calculations and assertions for Level specifications.


```python
# Level 12: Typed Measurement Packet
from fractions import Fraction


class MeasurementPacket:
    def __init__(self, measurand, indication, scale, uncertainty):
        self.measurand = measurand
        self.indication = indication
        self.value = Fraction(indication) * scale
        self.uncertainty = uncertainty


m = MeasurementPacket("Mass", 42, Fraction(1, 10), Fraction(1, 100))
assert m.value == Fraction(21, 5)
print(
    f"[Level 12 OK] Typed measurement packet verified: Value={m.value}, Uncertainty=+/-{m.uncertainty}"
)
```

## 2. Diagnostic Probes & Adversarial Falsifiers

Verify that un-admitted, unauthorized, or invalid operations are intercepted and rejected.


```python
# Falsifier test: Measurement without uncertainty bound must fail validation
def validate_measurement(packet):
    if packet.uncertainty is None:
        raise ValueError("[Falsifier OK] Measurement rejected: Missing uncertainty bound")


uncalibrated = MeasurementPacket("Voltage", 10, Fraction(1), None)
try:
    validate_measurement(uncalibrated)
except ValueError as e:
    print(e)
```
