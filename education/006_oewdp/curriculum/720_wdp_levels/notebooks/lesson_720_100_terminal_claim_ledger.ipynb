```python
# Level 100: Terminal SHA-256 Byte Digest & Git Object Framing
import hashlib


def git_blob_hash(content: bytes) -> str:
    header = f"blob {len(content)}\0".encode()
    return hashlib.sha1(header + content).hexdigest()


raw_data = b"WDP Level 100 Terminal Claim Ledger"
raw_sha256 = hashlib.sha256(raw_data).hexdigest()
git_sha1 = git_blob_hash(raw_data)

assert raw_sha256 != git_sha1, "Raw SHA256 and Git SHA1 must differ"
print(
    f"[Level 100 OK] Cryptographic digest framing verified:\n  Raw SHA256: {raw_sha256[:16]}...\n  Git SHA1:   {git_sha1[:16]}..."
)
```
