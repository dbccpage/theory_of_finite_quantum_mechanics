## 1. Primary Invariants & Primal/Dual Verification

Verify that the quotient certificate satisfies primal-dual equality and dual divergence freedom.


```python
from fractions import Fraction

# 4-cycle: vertices (0, 1, 2, 3), edges ((0,1), (1,2), (2,3), (0,3))
edges = ((0, 1), (1, 2), (2, 3), (0, 3))
cochain = (Fraction(1), Fraction(1), Fraction(1), Fraction(-1))
weights = (Fraction(1), Fraction(1), Fraction(1), Fraction(1))

# Candidate dual circulation around the cycle
dual = (Fraction(1), Fraction(1), Fraction(1), Fraction(-1))

# 1. Dual pairing sum(y_e * h_e)
dual_val = sum(y * h for y, h in zip(dual, cochain))
assert dual_val == Fraction(4)

# 2. Dual feasibility: |y_e| <= w_e
for y, w in zip(dual, weights):
    assert abs(y) <= w

# 3. Divergence-free at each vertex
div = {v: Fraction(0) for v in range(4)}
for (tail, head), y in zip(edges, dual):
    div[head] += y
    div[tail] -= y

assert all(d == 0 for d in div.values()), f"Nonzero divergence: {div}"
print("[Dual Certificate OK] Strong duality and zero divergence verified.")
```

## 3. Diagnostic Probes & Adversarial Falsifiers

Verify that a corrupted dual vector fails divergence-freedom.


```python
# Falsifier test: Corrupted dual with leakage at vertex 0
bad_dual = (Fraction(2), Fraction(1), Fraction(1), Fraction(-1))
bad_div = {v: Fraction(0) for v in range(4)}
for (tail, head), y in zip(edges, bad_dual):
    bad_div[head] += y
    bad_div[tail] -= y

assert any(d != 0 for d in bad_div.values()), "Failed to detect non-conserved circulation"
print(f"[Falsifier OK] Corrupted dual rejected: divergence {bad_div}")
```
