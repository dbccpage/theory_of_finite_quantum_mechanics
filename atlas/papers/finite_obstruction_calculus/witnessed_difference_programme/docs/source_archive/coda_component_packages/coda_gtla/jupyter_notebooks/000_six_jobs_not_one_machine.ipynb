# Six jobs, not one machine

Six companion papers do six different jobs: measure a leftover, record standing, move a picture, glue an overlap, read a shadow, compute a leftover vector. This notebook does not re-prove those papers. It runs the shipped finite evaluator and shows that the stages are different answers, not one machine.

This notebook uses the shipped calculator. Arithmetic is exact (fractions, not floating point).

Run from the package root, or from this `jupyter_notebooks/` folder.

> **Scope Note:** These notebooks demonstrate finite pedagogical examples and specific execution traces; they do not constitute mathematical proofs of universal claims.

```python
from pathlib import Path
import sys

p = Path.cwd().resolve()
if p.name == "jupyter_notebooks":
    root = p.parent
elif (p / "code" / "coda_gtla_core.py").is_file():
    root = p
elif (p / "coda_atlas" / "F005_coda" / "coda_gtla" / "code").is_dir():
    root = p / "coda_atlas" / "F005_coda" / "coda_gtla"
else:
    root = p
assert (root / "code").is_dir(), "run from the package root or jupyter_notebooks/"
if str(root / "code") not in sys.path:
    sys.path.insert(0, str(root / "code"))

from sympy import Matrix, Rational
from coda_gtla_core import (
    ProblemInstance,
    execute_master_normal_form,
    make_diag_span,
    FinitePresentation,
)
```

## Six desks

Write the six jobs in ordinary language:

1. Obstruction (FOC) — cheapest leftover after licensed linear repairs.
2. Standing (FDA) — a recorded pair, not a leftover number.
3. Transport (MUR) — a commuting square; collisions travel.
4. Comparison (OR) — pairs glue only when the overlap agrees.
5. Shadow (CS) — a blurry photo is not the original.
6. Leftover (A) — $x-Cx$ is a vector, not a yes.

The evaluator below calls some of those desks in order. Calling them in order does not make them one desk.

```python
d0 = Matrix([[1], [-1]])
d1 = Matrix([[1, 1]])
problem = ProblemInstance(
    problem_id="six_jobs",
    raw_data="licensed_exact",
    candidate_sieve=("cand1",),
    coarse_profile=(0, 1),
    fine_profile=(0, 1),
    projection_map={0: 0, 1: 1},
    local_sections={"U0": (Rational(0, 1),)},
    overlaps={("U0", "U0"): True},
    d0=d0,
    d1=d1,
    defect_cochain=Matrix([[1], [-1]]),  # in the image of d0
    fda_license_granted=True,
)
receipt = execute_master_normal_form(problem)
print("comparison sieve nonempty?", not receipt.r_stage.is_refused)
print("shadow refinement ok?", receipt.rho_stage.refinement_consistent)
print("overlaps glued?", receipt.d_stage.gluing_succeeded)
print("obstruction leftover Phi_1:", receipt.foc_stage.phi1_distance)
print("standing branch?", receipt.is_authorized())
print("leftover class?", receipt.is_anomalon())
assert receipt.r_stage.is_refused is False
assert receipt.rho_stage.refinement_consistent is True
assert receipt.d_stage.gluing_succeeded is True
assert receipt.foc_stage.phi1_distance == Rational(0, 1)
assert receipt.is_authorized() is True
assert receipt.is_anomalon() is False
```

## A span is not a leftover and not standing

The local Lean records a finite presentation and a diagonal span of cost zero. That cost is a number on a diagram. It does not measure $x-Cx$, and it does not inhabit a standing pair.

```python
P = FinitePresentation(
    name="two_points",
    carrier_size=2,
    diff_matrix=((Rational(0), Rational(1)), (Rational(1), Rational(0))),
)
diag = make_diag_span(P)
print("diagonal span cost:", diag.ext_cost)
assert diag.ext_cost == Rational(0, 1)
assert diag.left_map == (0, 1)
assert diag.right_map == (0, 1)
```
