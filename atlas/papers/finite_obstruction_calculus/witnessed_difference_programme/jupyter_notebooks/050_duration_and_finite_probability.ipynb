## Idea: two finite constructions, two explicit semantics

Witnessed duration is represented by ordered occurrences and positive rational
weights. A separate cochain classifier reports exact, closed-non-exact,
active-closure-failure, or exactness-undetermined status. Finite probability
uses multiplicity in an explicit tuple of occurrences.

Both are exact finite calculations. Neither makes an automatic claim about
universal physical time, stochastic propensity, or the Born rule.

## Derivation

For positive weights `w_i`, the bounded duration is their exact sum. For a
finite witness space `Omega` represented with repetition, the event proportion
is `number of admitted occurrences in A / number of occurrences in Omega`.
Disjoint additivity follows by partitioning those counted positions; it is not
imported from a measure on an unbounded sigma algebra.


```python
from witnessed_difference.cochains import FiniteComplex, d0
from witnessed_difference.duration import classify_duration, duration_sum, occurrence_chain
from witnessed_difference.notebook_support import display_json
from witnessed_difference.probability import FiniteWitnessSpace, disjoint_additivity, event_probability

chain = occurrence_chain(('registered', 'reaccessed', 'accepted'))
elapsed = duration_sum((1, '1/2', '3/2'))
interval = FiniteComplex(('t0', 't1'), (('t0', 't1'),))
duration_status = classify_duration(interval, d0(interval, {'t0': 0, 't1': 2}))
space = FiniteWitnessSpace(('red', 'red', 'blue', 'green'))
print(display_json({'chain': chain, 'duration': elapsed, 'classification': duration_status}))
print(display_json({'p_red': event_probability(space, frozenset({'red'})), 'additivity': disjoint_additivity(space, frozenset({'red'}), frozenset({'blue'}))}))
```

## Read the result

The duration sum is exactly `3`, not a floating approximation. The event
proportion for red is `1/2` because two of four recorded occurrences are red.
Repeated labels represent multiplicity; they are not silently deduplicated.


```python
from witnessed_difference.duration import ClockProfile, check_clock
from witnessed_difference.probability import conditional_probability

clock = ClockProfile('tick', 10, frozenset({0, 1, 2}), frozenset({0, 1, 2}), True)
conditional = conditional_probability(space, frozenset({'red'}), frozenset({'red', 'blue'}))
print(display_json({'clock_check': check_clock(clock), 'p_red_given_red_or_blue': conditional}))
```

## Exercise

Compute the proportion of `{blue, green}` and verify disjoint additivity for
the singleton events `{blue}` and `{green}`. Then try an event containing
`purple` and explain the typed refusal.


```python
answer = {
    'blue_or_green': event_probability(space, frozenset({'blue', 'green'})),
    'singleton_additivity': disjoint_additivity(space, frozenset({'blue'}), frozenset({'green'})),
    'outside_support': event_probability(space, frozenset({'purple'})),
}
print(display_json(answer))
```
