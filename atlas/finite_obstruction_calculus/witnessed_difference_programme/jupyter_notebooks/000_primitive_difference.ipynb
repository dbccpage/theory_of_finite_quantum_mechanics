## Idea: begin with less than a metric

`primitive_difference(a, b)` returns only **same** or **different**. It does
not assign distance, direction, probability, or physical meaning. A finite
carrier is supplied by the caller; the program never claims to derive that
carrier. Once labels are admitted, permutation enumeration lets us ask a
precise question: which candidate relations survive every relabelling?

This is the package's first recurring discipline: make the input structure
visible, exhaust the finite case when feasible, and keep the result narrower
than the motivating interpretation.

## Derivation

For an admitted binary labelling $\chi:X \to \{0,1\}$, define $d_\chi(a,b) = \chi(a) \mathbin{\oplus} \chi(b)$. Each vertex bit occurs twice in the triangle boundary, so its XOR is zero:

`d(a,b) XOR d(b,c) XOR d(a,c) = 0`.

The Boolean indicator of inequality between arbitrary labels does not satisfy this identity in general: three distinct labels give boundary `(1, 1, 1)`, whose XOR is 1. The identity holds strictly for binary-labelled carriers.

On a carrier with at least two points, a nontrivial permutation reverses any choice of total order, so no strict total order is invariant under *all* permutations.


```python
from witnessed_difference.ground import primitive_difference, triangle_boundary, triangle_identity
from witnessed_difference.notebook_support import display_json
from witnessed_difference.symmetry import exhaustive_relation_audit, fixed_points

binary_labels = ('red', 'red', 'blue')
three_labels = ('red', 'green', 'blue')
try:
    three_identity = triangle_identity(*three_labels)
except ValueError as err:
    three_identity = f"Refused: {err}"
report = {
    'binary_pairwise': [primitive_difference(binary_labels[i], binary_labels[j]) for i, j in ((0, 1), (1, 2), (0, 2))],
    'binary_triangle_boundary': triangle_boundary(*binary_labels),
    'binary_triangle_identity': triangle_identity(*binary_labels),
    'three_labels_boundary': triangle_boundary(*three_labels),
    'three_labels_identity': three_identity,
    'points_fixed_by_every_relabelling': fixed_points(3),
}
print(display_json(report))
print(display_json(exhaustive_relation_audit(3)))
```

## Read the result

For `n=3`, the audit checks all `2^(3*3)=512` directed binary relations and all
`3!=6` permutations. No point is fixed by every relabelling, illustrating the
distinguished-point obstruction. The returned invariant relations are finite witnesses,
not a classification theorem for every kind of structure. Notice also that
the exhaustive count depends on the declared convention that loops and
oriented pairs are allowed.


```python
from IPython.display import HTML, display
from witnessed_difference.notebook_support import slider_html

# In a normal browser this is a live dependency-free slider. The replay gate
# sets WDP_NOTEBOOK_STATIC=1 and records the deterministic fallback instead.
display(HTML(slider_html('Carrier size n', 1, 4, 3)))
summary = {
    str(n): {
        'invariant_relation_count': exhaustive_relation_audit(n)['invariant_relation_count'],
        'fixed_points': fixed_points(n),
    }
    for n in range(1, 4)
}
print(display_json({'finite_relabelling_audit': summary}))
```
