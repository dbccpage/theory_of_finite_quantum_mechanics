# What is left over?

You have a pair of numbers and a linear correction $C$. The leftover is the original pair minus the corrected pair: $\mathrm{residual}(C,x)=x-Cx$. That leftover is zero exactly when the correction did not change the pair.

This notebook uses the shipped calculator. Arithmetic is exact (fractions, not floating point).

Run from the package root, or from this `jupyter_notebooks/` folder.

```python
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
root = cwd.parent if cwd.name == "jupyter_notebooks" else cwd
code_dir = (root / "code").resolve()
assert (code_dir / "coda_a_core.py").is_file(), "Run from this package root or jupyter_notebooks/."
sys.path[:] = [str(code_dir)] + [entry for entry in sys.path if entry != str(code_dir)]
import coda_a_core as companion
assert Path(companion.__file__).resolve() == code_dir / "coda_a_core.py", "An ambient module shadowed the local companion."
import cochain_presentation as cochain
assert Path(cochain.__file__).resolve() == code_dir / "cochain_presentation.py", "An ambient solver shadowed the local solver."

from sympy import Matrix
from coda_a_core import residual
```

## A picture in the plane

Let $C$ keep the first number and throw away the second: $C(a,b)=(a,0)$. The pair $(3,4)$ becomes $(3,0)$. What is left over is $(3,4)-(3,0)=(0,4)$.

```python
C = Matrix([[1, 0], [0, 0]])
x = Matrix([[3], [4]])
corrected = C * x
leftover = residual(C, x)
print("original:", list(x))
print("after C:", list(corrected))
print("leftover:", list(leftover))
assert leftover == x - corrected
assert leftover == Matrix([[0], [4]])
```

## Zero leftover means the pair was already fixed

$(3,0)$ is already on the first axis. $C$ does not change it, so the leftover is $(0,0)$. The leftover of $(3,4)$ is not zero, because $C$ did change it. That is the whole comparison: leftover zero if and only if $Cx=x$. It does not decide what to do next.

```python
fixed = Matrix([[3], [0]])
print("leftover of (3,0):", list(residual(C, fixed)))
print("(3,0) already equal to C of it?", C * fixed == fixed)
print("(3,4) already equal to C of it?", C * x == x)
assert residual(C, fixed).is_zero_matrix is True
assert residual(C, x).is_zero_matrix is False
assert (residual(C, x).is_zero_matrix) == (C * x == x)
assert (residual(C, fixed).is_zero_matrix) == (C * fixed == fixed)
```
