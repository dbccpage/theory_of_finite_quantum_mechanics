# Moving leftovers with a linear map

If you stretch both the original pair and the correction with a linear map $T$, and $T$ does not scramble the correction ($TC=DT$), then leftovers stretch by $T$ too. Two pairs have the same leftover exactly when their difference is something $C$ can already produce.

This notebook uses the shipped calculator. Arithmetic is exact (fractions, not floating point).

Run from the package root, or from this `jupyter_notebooks/` folder.

```python
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
root = cwd.parent if cwd.name == "jupyter_notebooks" else cwd
code_dir = (root / "code").resolve()
assert (code_dir / "coda_a_core.py").is_file(), "Run from this package root or jupyter_notebooks/."
sys.path[:] = [str(code_dir)] + [entry for entry in sys.path if entry != str(code_dir)]
import coda_a_core as companion
assert Path(companion.__file__).resolve() == code_dir / "coda_a_core.py", "An ambient module shadowed the local companion."
import cochain_presentation as cochain
assert Path(cochain.__file__).resolve() == code_dir / "cochain_presentation.py", "An ambient solver shadowed the local solver."

from sympy import Matrix
from coda_a_core import difference_in_image, maps_commute, residual
```

## Stretching by two commutes

Let $C$ keep the first coordinate. Let $T$ multiply both coordinates by $2$. Then $T$ after $C$ is the same as $C$ after $T$. The leftover $(0,4)$ stretches to $(0,8)$, which is the leftover of the stretched pair $(6,8)$.

```python
C = Matrix([[1, 0], [0, 0]])
x = Matrix([[3], [4]])
T = Matrix([[2, 0], [0, 2]])
print("T C = C T?", maps_commute(T, C, C))
print("T of leftover:", list(T * residual(C, x)))
print("leftover of T x:", list(residual(C, T * x)))
assert maps_commute(T, C, C) is True
assert T * residual(C, x) == residual(C, T * x)
```

## Swapping coordinates does not commute

Let $S(a,b)=(b,a)$. Then $S$ after $C$ is not $C$ after $S$. Leftovers do not travel with $S$: $S$ of $(0,4)$ is $(4,0)$, while the leftover of $(4,3)$ is $(0,3)$.

```python
S = Matrix([[0, 1], [1, 0]])
print("S C = C S?", maps_commute(S, C, C))
print("S of leftover:", list(S * residual(C, x)))
print("leftover of S x:", list(residual(C, S * x)))
assert maps_commute(S, C, C) is False
assert S * residual(C, x) != residual(C, S * x)
```

## Same leftover, same class

$(3,4)$ and $(5,4)$ differ by $(-2,0)$, which $C$ can produce. They have the same leftover $(0,4)$. $(3,4)$ and $(3,5)$ differ in the second coordinate, which $C$ cannot produce. Their leftovers differ. A leftover class is not a decision.

```python
y_same = Matrix([[5], [4]])
y_diff = Matrix([[3], [5]])
print("same leftover as (5,4)?", residual(C, x) == residual(C, y_same))
print("difference in image of C?", difference_in_image(C, x, y_same))
print("same leftover as (3,5)?", residual(C, x) == residual(C, y_diff))
print("that difference in image of C?", difference_in_image(C, x, y_diff))
assert difference_in_image(C, x, y_same) is True
assert residual(C, x) == residual(C, y_same)
assert difference_in_image(C, x, y_diff) is False
assert residual(C, x) != residual(C, y_diff)
```
