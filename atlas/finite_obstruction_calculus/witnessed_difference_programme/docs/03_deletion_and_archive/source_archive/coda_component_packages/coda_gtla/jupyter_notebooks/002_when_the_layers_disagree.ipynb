# When the layers disagree

The six jobs can disagree. An empty candidate list refuses before leftover is computed. Two records can share leftover numbers and disagree on standing. A closed leftover is not a standing branch. Disagreement is data. Averaging it away is not a job of this calculator.

This notebook uses the shipped calculator. Arithmetic is exact (fractions, not floating point).

Run from the package root, or from this `jupyter_notebooks/` folder.

> **Scope Note:** These notebooks demonstrate finite pedagogical examples and specific execution traces; they do not constitute mathematical proofs of universal claims.

```python
from pathlib import Path
import sys

p = Path.cwd().resolve()
if p.name == "jupyter_notebooks":
    root = p.parent
elif (p / "code" / "coda_gtla_core.py").is_file():
    root = p
elif (p / "coda_atlas" / "F005_coda" / "coda_gtla" / "code").is_dir():
    root = p / "coda_atlas" / "F005_coda" / "coda_gtla"
else:
    root = p
assert (root / "code").is_dir(), "run from the package root or jupyter_notebooks/"
if str(root / "code") not in sys.path:
    sys.path.insert(0, str(root / "code"))

from sympy import Matrix, Rational
from coda_gtla_core import (
    ProblemInstance,
    execute_master_normal_form,
    observation_collision_countermodel,
    reproduce_master_normal_form_example,
)
```

## Three outcomes, three constructors

The shipped example runs three problems: a licensed exact repair, a closed leftover, and an empty candidate list. They land in three different constructors. No constructor is a blend of the others.

```python
demo = reproduce_master_normal_form_example()
print("licensed exact:", demo["case_1_auth"])
print("closed leftover:", demo["case_2_anomalon"])
print("empty list:", demo["case_3_refusal"])
assert demo["case_1_auth"]["is_authorized"] is True
assert demo["case_2_anomalon"]["is_anomalon"] is True
assert demo["case_3_refusal"]["is_refused"] is True
assert demo["partition_disjoint"] is True
```

## Same leftover numbers, different standing

Two problems can share the same cochain and the same leftover numbers, and still disagree on the licence flag. Looking at $\Phi_1$ cannot tell them apart. That is the observation-collision pattern, run here on leftover numbers.

```python
report = observation_collision_countermodel()
print(report)
assert report["observation_identical"] is True
assert report["s1_authorized"] is True
assert report["s2_authorized"] is False
assert report["authority_separated"] is True
```

## Open leftover is not closed leftover

If the next-stage map sees the defect, the leftover is open. The calculator refuses. That is not the closed-leftover constructor, and it is not standing.

```python
open_leftover = ProblemInstance(
    problem_id="open_leftover",
    raw_data="nonclosed",
    candidate_sieve=("cand1",),
    coarse_profile=(0, 1),
    fine_profile=(0, 1),
    projection_map={0: 0, 1: 1},
    local_sections={"U0": (Rational(0, 1),)},
    overlaps={("U0", "U0"): True},
    d0=Matrix([[1], [1]]),
    d1=Matrix([[1, -1]]),
    defect_cochain=Matrix([[2], [0]]),
    fda_license_granted=True,
)
rec = execute_master_normal_form(open_leftover)
print("Gamma_2:", rec.terminal_outcome.gamma2_distance)
print("kind:", rec.terminal_outcome.kind.value)
assert rec.terminal_outcome.gamma2_distance > Rational(0, 1)
assert rec.is_refused() is True
assert rec.is_authorized() is False
assert rec.is_anomalon() is False
```
