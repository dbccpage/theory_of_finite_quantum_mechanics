## §6.2 open-chain TFIM: $\nu_e = 2|J|$

For the transverse-field Ising chain in a product $|+\rangle$ state, the paper's
§6.2 computation gives $\nu_e = 2|J|$ on every bond. The replay recomputes each
edge rate from the reduced derivative and reports the largest deviation.

```python
replay = tfim_normal_rate_example(num_sites=4, coupling=1.0, field=0.5)

print(f"sites            : {replay.num_sites}")
print(f"coupling J       : {replay.coupling}")
print(f"transverse field : {replay.field}")
print()
print("edge            nu_e")
print("-" * 28)
for edge, rate in replay.edge_rates:
    print(f"{edge[0]}--{edge[1]:<8}  {rate:.12f}")
print()
print(f"expected 2|J|        : {replay.expected_rate}")
print(f"max deviation        : {replay.maximum_rate_error:.3e}")
print(f"all edges active     : {replay.all_edges_active}")
```

## The incident interaction norm $g$

$g = \max_v \sum_{X \ni v} \lVert h_X \rVert_{\mathrm{op}}$ is the *full incident*
norm — the quantity Theorem 5.1's $24g^2$ term is stated in. It is computed from
the Hamiltonian terms, never estimated from the coupling constant.

```python
print(f"full incident g      : {replay.full_incident_g:.12f}")
print(f"paper bound 2|J|+|h| : {replay.paper_g_upper_bound:.12f}")
print(f"within paper bound   : {replay.g_within_paper_bound}")
```
