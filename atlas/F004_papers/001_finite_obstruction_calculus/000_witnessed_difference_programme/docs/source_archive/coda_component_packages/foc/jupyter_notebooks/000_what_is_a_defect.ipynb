# What is a defect?

You have a list of numbers — a vector. You are allowed to add or subtract along certain directions. Those directions are the *repairs already licensed*. Whatever you cannot cancel is the *leftover*, the defect.

This notebook uses the shipped calculator. Arithmetic is exact (fractions, not floating point).

## The setup, in words

Work in the plane $\mathbb{R}^2$. Licensed repairs are “add any multiple of $(1,0)$”: you may freely correct the first coordinate, and you may not touch the second. The next-defect map is zero, so nothing is sent one degree up.

In code that is one packet: repair matrix `d0 = [[1], [0]]`, next-defect matrix `d1 = [[0, 0]]`, no second-stage repairs, and weight $1$ on every coordinate.

```python
from sympy import Matrix, Rational
from finite_obstruction import FiniteObstructionPacket, diagnose

packet = FiniteObstructionPacket.create(
    d0=[[1], [0]],
    d1=[[0, 0]],
    a2=Matrix.zeros(1, 0),
    w1=[1, 1],
    w2=[1],
)
print("packet accepted")
```

## One repairable example

The vector $(3,0)$ lies on the repair line. Subtract $3\cdot(1,0)$ and nothing is left. The leftover is $0$: an exact repair.

```python
exact = diagnose(packet, [3, 0])
print("diagnostic:", exact.diagnostic.value)
print("leftover Phi1:", exact.phi1.value)
print("next-stage leftover Gamma2:", exact.gamma2.value)
assert exact.diagnostic.value == "ExactRepair"
assert exact.phi1.value == Rational(0)
```

## One leftover residual

The vector $(0,3)$ cannot be cancelled. No multiple of $(1,0)$ touches the second slot. Measured by ordinary $\ell^1$ (sum of absolute values), the leftover is $3$.

Because the next-defect map is zero, that leftover produces nothing one degree up. The calculator therefore reports a *coherent residual*: a real leftover that does not leak to the next stage.

```python
left = diagnose(packet, [0, 3])
print("diagnostic:", left.diagnostic.value)
print("leftover Phi1:", left.phi1.value)
print("next-stage leftover Gamma2:", left.gamma2.value)
assert left.diagnostic.value == "CoherentResidual"
assert left.phi1.value == Rational(3)
assert left.gamma2.value == Rational(0)
```
