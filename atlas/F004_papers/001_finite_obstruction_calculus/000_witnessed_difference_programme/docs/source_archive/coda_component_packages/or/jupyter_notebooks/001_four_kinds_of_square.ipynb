# Four kinds of square (CODA-OR)

A comparison square has a leftover $\delta$. After licensed repairs $d_0$, and after looking one degree up with $d_1$, there are four labels (OR-D2, OR-T4).

| Label | In words |
|---|---|
| Strict commute | The leftover is already $0$. |
| Exact repair | A licensed repair cancels it ($d_0 \alpha = \delta$). |
| Closed leftover | It cannot be cancelled, but the next stage is $0$ ($d_1 \delta = 0$). |
| Open leftover | It still shows one degree up ($d_1 \delta \neq 0$). |

Run from the package root, or from this `jupyter_notebooks/` folder.

```python
from pathlib import Path
import sys

p = Path.cwd().resolve()
if p.name == "jupyter_notebooks":
    root = p.parent
elif (p / "code" / "coda_or_core.py").is_file():
    root = p
elif (p / "coda_atlas" / "F005_coda" / "or" / "code").is_dir():
    root = p / "coda_atlas" / "F005_coda" / "or"
else:
    root = p

assert (root / "code").is_dir(), "run from package root or jupyter_notebooks/"
if str(root / "code") not in sys.path:
    sys.path.insert(0, str(root / "code"))

from sympy import Rational, Matrix
from coda_or_core import reproduce_section_6a_worked_example, SquareStatus
print("CODA-OR square classifier loaded.")
```

```python
rows = reproduce_section_6a_worked_example()
assert len(rows) == 4, "Expected 4 rows from Section 6A replay"

# Print results
print(f"{'Square':<10} | {'Status':<24} | {'Dist':<6} | {'Closed':<6} | {'Exact':<6} | {'Boundary d1*delta':<18}")
print("-" * 80)
for i, row in enumerate(rows, start=1):
    print(f"{i:<10} | {row.status.value:<24} | {str(row.quotient_distance):<6} | {str(row.is_closed):<6} | {str(row.is_exact):<6} | {str(list(row.boundary_image)):<18}")

# Assertions for each square
# Square 1: Strict commute
assert rows[0].status == SquareStatus.STRICT_COMMUTE
assert rows[0].quotient_distance == Rational(0, 1)
assert rows[0].is_closed is True
assert rows[0].is_exact is True

# Square 2: Exact repair (WeakCommuteCertified)
assert rows[1].status == SquareStatus.WEAK_COMMUTE_CERTIFIED
assert rows[1].quotient_distance == Rational(0, 1)
assert rows[1].is_closed is True
assert rows[1].is_exact is True
# Verify exact repair identity d0 * alpha = delta (d0=[[1]], alpha=[[3]], delta=[[3]])
assert rows[1].repair_2cell == Matrix([[3]])

# Square 3: Closed diagram tear (H^1DiagramTear)
assert rows[2].status == SquareStatus.H1_DIAGRAM_TEAR
assert rows[2].quotient_distance == Rational(3, 1)
assert rows[2].is_closed is True
assert rows[2].is_exact is False

# Square 4: Open diagram obstruction
assert rows[3].status == SquareStatus.OPEN_DIAGRAM_OBSTRUCTION
assert rows[3].quotient_distance == Rational(3, 1)
assert rows[3].is_closed is False
assert rows[3].is_exact is False
assert rows[3].boundary_image == Matrix([[3]])

print("\nAll Section 6A mathematical invariants asserted and verified.")
```
