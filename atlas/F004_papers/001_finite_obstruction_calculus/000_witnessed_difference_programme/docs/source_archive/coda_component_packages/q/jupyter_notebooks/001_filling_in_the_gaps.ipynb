# Filling in the gaps

No rational squares to $2$. You can still write a list of rationals that get closer and closer to one another, and closer to the missing square root. Completing the rationals adds that missing limit. The original rationals remain dense: every completed point can be squeezed between rationals.

This notebook uses nested exact enclosures. Arithmetic is exact (fractions, not floating point).

Run from the package root, or from this `jupyter_notebooks/` folder.

```python
from pathlib import Path
import sys

p = Path.cwd().resolve()
if p.name == "jupyter_notebooks":
    root = p.parent
elif (p / "code" / "coda_q_core.py").is_file():
    root = p
elif (p / "coda_atlas" / "F005_coda" / "q" / "code").is_dir():
    root = p / "coda_atlas" / "F005_coda" / "q"
elif (p / "q" / "code").is_dir():
    root = p / "q"
else:
    root = p
assert (root / "code").is_dir(), "run from the package root or jupyter_notebooks/"
if str(root / "code") not in sys.path:
    sys.path.insert(0, str(root / "code"))

from fractions import Fraction
from coda_q_core import (
    enclosure_endpoint_sequence,
    is_cauchy_from,
    nested_sqrt2_enclosures,
    sequences_close_from,
)
```

## Nested intervals

Each bisection keeps $\sqrt{2}$ and halves the width. The lower ends only go up; the upper ends only go down. After $n$ steps the width is $2^{-n}$.

```python
encs = nested_sqrt2_enclosures(5)
for k, enc in enumerate(encs):
    print(f"n={k}: [{enc.lower}, {enc.upper}] width={enc.upper - enc.lower}")
    assert enc.verifies_sqrt2() is True
assert encs[0].upper - encs[0].lower == Fraction(1)
assert encs[5].upper - encs[5].lower == Fraction(1, 32)
assert encs[5].lower >= encs[0].lower
assert encs[5].upper <= encs[0].upper
```

## Later terms sit close together

A sequence is Cauchy when, from some index on, every pair of terms is closer than a chosen tolerance. The lower endpoints of the nested intervals have that property: once the width is smaller than the tolerance, later lowers cannot jump farther than that width. The calculator checks this on a finite prefix. That is a modulus check, not a construction of every real.

```python
lowers = enclosure_endpoint_sequence(6, "lower")
uppers = enclosure_endpoint_sequence(6, "upper")
mids = enclosure_endpoint_sequence(6, "midpoint")
print("lowers:", list(lowers.terms))
print("Cauchy from n=2 at tolerance 1/2?", is_cauchy_from(lowers, 2, Fraction(1, 2)))
print("lowers vs midpoints from n=3 at tolerance 1/2?", sequences_close_from(lowers, mids, 3, Fraction(1, 2)))
assert is_cauchy_from(lowers, 2, Fraction(1, 2)) is True
assert is_cauchy_from(uppers, 2, Fraction(1, 2)) is True
assert sequences_close_from(lowers, mids, 3, Fraction(1, 2)) is True
```

## The original rationals stay in the picture

Every enclosure is a pair of rationals. The completed point $\sqrt{2}$ is squeezed between them. That is density in one example: you never leave the rationals when you write the ends, and the gap can be made as small as you like. Filling the gap adds a limit. It does not print a rational whose square is $2$ — there is none.

```python
enc = encs[-1]
print("after 5 steps, lower squared:", enc.lower ** 2)
print("after 5 steps, upper squared:", enc.upper ** 2)
print("no endpoint squares to 2:", enc.lower ** 2 != 2 and enc.upper ** 2 != 2)
assert enc.lower ** 2 < 2 < enc.upper ** 2
assert enc.lower ** 2 != 2
assert enc.upper ** 2 != 2
```
