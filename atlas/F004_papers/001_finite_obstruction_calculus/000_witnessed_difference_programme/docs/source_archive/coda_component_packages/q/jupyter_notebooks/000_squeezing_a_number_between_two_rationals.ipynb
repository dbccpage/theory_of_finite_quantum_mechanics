# Squeezing a number between two rationals

You know $\sqrt{2}$ sits somewhere between $1$ and $2$, because $1^2=1$ is too small and $2^2=4$ is too big. The midpoint is $3/2$. That midpoint is a rational you can write down. It is not $\sqrt{2}$.

This notebook uses the shipped calculator. Arithmetic is exact (fractions, not floating point).

Run from the package root, or from this `jupyter_notebooks/` folder.

```python
from pathlib import Path
import sys

p = Path.cwd().resolve()
if p.name == "jupyter_notebooks":
    root = p.parent
elif (p / "code" / "coda_q_core.py").is_file():
    root = p
elif (p / "coda_atlas" / "F005_coda" / "q" / "code").is_dir():
    root = p / "coda_atlas" / "F005_coda" / "q"
elif (p / "q" / "code").is_dir():
    root = p / "q"
else:
    root = p
assert (root / "code").is_dir(), "run from the package root or jupyter_notebooks/"
if str(root / "code") not in sys.path:
    sys.path.insert(0, str(root / "code"))

from fractions import Fraction
from coda_q_core import enclosure_error, enclosure_error_bound, midpoint, sqrt_step
from continuum import sqrt2_enclosure
```

## Start with $[1,2]$

The unknown sits in the interval. The calculator records the two rational ends, the midpoint, and half the width.

```python
enc0 = sqrt2_enclosure(0)
print("interval:", enc0.lower, "to", enc0.upper)
print("midpoint:", enc0.midpoint)
print("half-width:", enc0.error_bound)
print("squares bracket 2?", enc0.verifies_sqrt2())
assert enc0.lower == Fraction(1)
assert enc0.upper == Fraction(2)
assert enc0.midpoint == Fraction(3, 2)
assert enc0.error_bound == Fraction(1, 2)
assert enc0.verifies_sqrt2() is True
```

## The midpoint is not the number

$(3/2)^2=9/4$, which is larger than $2$. So $\sqrt{2}$ is to the left of $3/2$. Theorem Q-T4 says every point of $[1,2]$ is at most $1/2$ from the midpoint. That bound is attained at the endpoints. It is a bound, not an identification.

```python
print("(3/2)^2 =", Fraction(3, 2) ** 2)
print("error at left end:", enclosure_error(1, 2, 1))
print("error at midpoint:", enclosure_error(1, 2, Fraction(3, 2)))
print("half-width:", enclosure_error_bound(1, 2))
assert Fraction(3, 2) ** 2 == Fraction(9, 4)
assert Fraction(9, 4) > 2
assert enclosure_error(1, 2, 1) == enclosure_error_bound(1, 2)
assert enclosure_error(1, 2, Fraction(3, 2)) == 0
assert midpoint(1, 2) != Fraction(0)  # a written midpoint is still a rational
```

## One bisection step

Compare the midpoint square with $2$. If it is already larger than $2$, keep the left half. The new interval still contains $\sqrt{2}$, and it is half as wide.

```python
lo, hi = sqrt_step(1, 2)
print("after one step:", lo, "to", hi)
print("new width:", hi - lo)
print("still brackets 2?", lo ** 2 <= 2 <= hi ** 2)
assert (lo, hi) == (Fraction(1), Fraction(3, 2))
assert hi - lo == Fraction(1, 2)
assert lo ** 2 <= 2 <= hi ** 2

lo2, hi2 = sqrt_step(lo, hi)
print("after two steps:", lo2, "to", hi2)
assert (lo2, hi2) == (Fraction(5, 4), Fraction(3, 2))
assert hi2 - lo2 == Fraction(1, 4)
assert lo2 ** 2 <= 2 <= hi2 ** 2
```
