# A leftover is not a yes

A closed leftover is a class in a quotient: the defect is not an exact repair, and the next-stage leftover is zero. That class is not standing. Zero leftover is not standing either, unless a separate licence flag is already set — and even then, the flag is a declared stand-in, not a proof of standing.

This notebook uses the shipped calculator. Arithmetic is exact (fractions, not floating point).

Run from the package root, or from this `jupyter_notebooks/` folder.

> **Scope Note:** These notebooks demonstrate finite pedagogical examples and specific execution traces; they do not constitute mathematical proofs of universal claims.

```python
from pathlib import Path
import sys

p = Path.cwd().resolve()
if p.name == "jupyter_notebooks":
    root = p.parent
elif (p / "code" / "coda_gtla_core.py").is_file():
    root = p
elif (p / "coda_atlas" / "F005_coda" / "coda_gtla" / "code").is_dir():
    root = p / "coda_atlas" / "F005_coda" / "coda_gtla"
else:
    root = p
assert (root / "code").is_dir(), "run from the package root or jupyter_notebooks/"
if str(root / "code") not in sys.path:
    sys.path.insert(0, str(root / "code"))

from sympy import Matrix, Rational
from coda_gtla_core import ProblemInstance, TerminalKind, execute_master_normal_form
```

## A closed leftover is a class, not a yes

Take a defect $(1,1)$ with no licensed repair ($d_0=0$) and no next-stage map ($d_1=0$). The first-stage leftover is $2$. The next-stage leftover is $0$. The calculator records a leftover class. It does not record standing.

```python
closed = ProblemInstance(
    problem_id="closed_leftover",
    raw_data="leftover",
    candidate_sieve=("cand1",),
    coarse_profile=(0, 1),
    fine_profile=(0, 1),
    projection_map={0: 0, 1: 1},
    local_sections={"U0": (Rational(1, 2),)},
    overlaps={("U0", "U0"): True},
    d0=Matrix([[0], [0]]),
    d1=Matrix([[0, 0]]),
    defect_cochain=Matrix([[1], [1]]),
    fda_license_granted=False,
)
rec = execute_master_normal_form(closed)
print("Phi_1:", rec.terminal_outcome.phi1_distance)
print("kind:", rec.terminal_outcome.kind.value)
print("leftover class?", rec.is_anomalon())
print("standing branch?", rec.is_authorized())
assert rec.terminal_outcome.phi1_distance == Rational(2, 1)
assert rec.terminal_outcome.kind == TerminalKind.CLOSED_RESIDUAL
assert rec.is_anomalon() is True
assert rec.is_authorized() is False
assert rec.is_refused() is False
```

## Zero leftover is not a yes either

Now take a defect that *is* an exact repair, so leftover $0$, but leave the licence flag unset. The obstruction desk says zero. The standing desk still says no.

```python
zero = ProblemInstance(
    problem_id="zero_unlicensed",
    raw_data="zero",
    candidate_sieve=("cand1",),
    coarse_profile=(0, 1),
    fine_profile=(0, 1),
    projection_map={0: 0, 1: 1},
    local_sections={"U0": (Rational(0, 1),)},
    overlaps={("U0", "U0"): True},
    d0=Matrix([[1], [-1]]),
    d1=Matrix([[1, 1]]),
    defect_cochain=Matrix([[1], [-1]]),
    fda_license_granted=False,
)
rec0 = execute_master_normal_form(zero)
print("Phi_1:", rec0.terminal_outcome.phi1_distance)
print("kind:", rec0.terminal_outcome.kind.value)
print("standing branch?", rec0.is_authorized())
assert rec0.terminal_outcome.phi1_distance == Rational(0, 1)
assert rec0.is_authorized() is False
assert rec0.is_refused() is True
assert rec0.is_anomalon() is False
```
