## Idea: same word, different mathematical types

The monograph discusses three finite notions under the word *observer*: a
quotient presentation, an operational transition/report packet, and a richer
observer object with defect, repair, gauge, transformation, and terminalizer
fields. They are not aliases. A bridge is data that declares how one
representation is interpreted as another.

Refusing a missing or mismatched bridge is part of the result, not an
inconvenience to suppress.

## Derivation

An operational packet separates a state's externally reported symbol from its
profile. If two profile-distinct states emit the same report, the finite
separation check refuses the packet. This is an injectivity condition only on
the declared state/report/profile table; it is not a theory of subjective
observation.


```python
from witnessed_difference.notebook_support import display_json
from witnessed_difference.observer import ObserverBridge, QuotientObserver, adapt_quotient

quotient = QuotientObserver(('a', 'b'), (frozenset({'a'}), frozenset({'b'})), 'equality')
bridge = ObserverBridge('quotient', 'local-state', 'delta', 'd0', None, 'gauge', 'transform', 'accept')
print(display_json({'without_bridge': adapt_quotient(quotient, None)}))
print(display_json({'with_bridge': adapt_quotient(quotient, bridge)}))
```

## Read the result

The first outcome refuses an implicit conversion. The second constructs an
`ObserverObject` because every required bridge field is present. The optional
closure field remains `None`; this makes the absence visible without confusing
it with a required-field failure.


```python
from witnessed_difference.observer import OperationalObserverPacket, validate_report_profile_separation

good = OperationalObserverPacket(
    ('s0', 's1'), 's0', ('tick',), (('s0', 'tick', 's1'), ('s1', 'tick', 's0')),
    (('s0', 'quiet'), ('s1', 'signal')),
    (('s0', 'baseline'), ('s1', 'changed')),
)
collapsed = OperationalObserverPacket(
    ('s0', 's1'), 's0', ('tick',), (('s0', 'tick', 's1'), ('s1', 'tick', 's0')),
    (('s0', 'same-report'), ('s1', 'same-report')),
    (('s0', 'baseline'), ('s1', 'changed')),
)
print(display_json({'separated': validate_report_profile_separation(good), 'collapsed': validate_report_profile_separation(collapsed)}))
```
