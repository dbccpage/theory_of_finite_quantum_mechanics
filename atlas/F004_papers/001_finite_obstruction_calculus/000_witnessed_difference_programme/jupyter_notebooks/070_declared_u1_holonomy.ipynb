## Idea: declared quantum-flavoured data are still declared data

A finite `U1Connection` assigns unit complex phases to admitted oriented
edges. Reversing an edge uses the complex conjugate. Multiplication around a
closed path gives holonomy. A vertex gauge assignment transforms edge phases
while leaving closed-loop holonomy exactly unchanged over rational complex phases.

The module labels this as a finite declared calculation. It deliberately
refuses the inference to a complete reconstruction of quantum theory.

## Derivation

For gauge values `g_v` and edge phase `U_uv`, the transformed phase is
`conj(g_u) * U_uv * g_v`. Around a closed path, adjacent `g_v` and
`conj(g_v)` factors cancel, and the initial vertex factors cancel at closure.
This calculation assumes the connection, path, and gauge carrier supplied by
the caller.


```python
from witnessed_difference.notebook_support import display_json
from witnessed_difference.quantum import gauge_transform, holonomy, make_connection

made = make_connection(
    ('a', 'b', 'c'),
    (('a', 'b', 1j), ('b', 'c', 1j), ('a', 'c', 1 + 0j)),
)
connection = made.value
loop = ('a', 'b', 'c', 'a')
before = holonomy(connection, loop)
transformed = gauge_transform(connection, {'a': 1 + 0j, 'b': -1 + 0j, 'c': 1j}).value
after = holonomy(transformed, loop)
print(display_json({'holonomy_before': before, 'holonomy_after': after}))
```

## Read the result

Both holonomies are exactly `-1`. Real and imaginary coordinates are
represented by rational numbers, and unit modulus is checked exactly. This
supports the displayed finite calculation; it does not cover arbitrary
algebraic phases. The image is a visualization, not additional evidence.


```python
from witnessed_difference.quantum import QuantumEvidence, holonomy, supports_quantum_reconstruction

open_path = holonomy(connection, ('a', 'b', 'c'))
scope_firewall = supports_quantum_reconstruction(QuantumEvidence.DECLARED_U1_CALCULATION)
print(display_json({'open_path': open_path, 'reconstruction_request': scope_firewall}))
```
