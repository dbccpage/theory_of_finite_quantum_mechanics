## Idea: attack the finite contract, then restore custody

The package's mutation gate copies a source tree into scratch space, applies one
named textual change, imports from that mutant copy, and runs the contract tests
expected to fail. After every mutant it removes scratch state and verifies
that the pristine source-tree digest is unchanged. A killed mutant demonstrates
that a particular injected defect was detected by particular tests.

It does **not** prove absence of all defects. Mutation evidence, branch
coverage, notebook replay, Lean elaboration, and prose review answer different
questions and must remain separately reported.

## Derivation

Let `H(S)` be the canonical source-tree hash and `M_i` a declared copy-on-write
mutation. The recovery invariant is `H(S_before) = H(S_after)`; the kill
condition is a nonzero test result from the mutant import path with the expected
contract node IDs implicated. Neither condition licenses replacing the source
claim ledger with a single green badge.


```python
import hashlib
from witnessed_difference.notebook_support import display_json

pristine = b'def authority_effect(): return "None"\n'
mutant = pristine.replace(b'"None"', b'"Promote"')
restored = bytes(pristine)
digest = lambda payload: hashlib.sha256(payload).hexdigest()
print(display_json({
    'copy_on_write': pristine != mutant,
    'mutant_digest_differs': digest(pristine) != digest(mutant),
    'pristine_recovered': digest(pristine) == digest(restored),
    'source_digest': digest(pristine),
}))
```

## Read the result

The in-memory mutant differs and the restored bytes match the pristine digest.
This tiny cell illustrates the custody invariant only. The release mutation
receipt is produced by `scripts/check_mutations.py`, which executes named real
contract tests against each scratch copy.
