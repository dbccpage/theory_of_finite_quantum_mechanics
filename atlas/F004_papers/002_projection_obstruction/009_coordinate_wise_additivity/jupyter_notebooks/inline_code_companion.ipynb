## Section 2.4 Worked Replay: Two-Edge Orbit-Weighted vs Unweighted Total

Verifying that graph invariance alone permits orbit-weighted $\ell^1$, while unweighted global $\ell^1$ requires edge-transitivity or explicit uniformity.


```python
d1 = (3, -4)
d2 = (1, 2)
w1, w2 = 1, 2

def l1(x):
    return sum(abs(t) for t in x)

orbit_weighted = w1 * l1(d1) + w2 * l1(d2)
unweighted = l1(d1) + l1(d2)

print("orbit_weighted:", orbit_weighted)
print("unweighted:", unweighted)

assert orbit_weighted == 13
assert unweighted == 10
print("Replay verified successfully!")
```
